```python
import numpy as np
import matplotlib.pyplot as plt
import segyio
import segyio.tools
from scipy.signal import correlate
from scipy.stats import spearmanr

# Класс из предыдущей работы (с небольшими дополнениями)
class RickerWavelet:
    def __init__(self, dominant_freq=30, dt=0.001, length=0.2):
        self.freq = dominant_freq
        self.dt = dt
        self.length = length
        self.time = np.arange(-length/2, length/2, dt)
        tau = np.pi * self.freq * self.time
        self.wavelet = (1 - 2 * tau**2) * np.exp(-tau**2)
    
    def get_wavelet(self):
        return self.time, self.wavelet

class ReflectionCoefficients:
    def __init__(self, depths, velocities, densities):
        self.depths = np.array(depths)
        self.velocities = np.array(velocities)
        self.densities = np.array(densities)
        self.impedances = self.densities * self.velocities
        
        # Коэффициенты отражения на каждой границе
        self.coeffs = []
        for i in range(len(self.depths)):
            Z_up = self.impedances[i]
            Z_down = self.impedances[i+1]
            R = (Z_down - Z_up) / (Z_down + Z_up)
            self.coeffs.append(R)
        self.coeffs = np.array(self.coeffs)
        
        # Времена прихода (двусторонние)
        self.times = []
        cum_time = 0
        prev_depth = 0
        for i in range(len(self.depths)):
            thickness = self.depths[i] - prev_depth
            v = self.velocities[i]
            cum_time += 2 * thickness / v
            self.times.append(cum_time)
            prev_depth = self.depths[i]
        self.times = np.array(self.times)
    
    def get_reflectivity_series(self, dt, total_time):
        ns = int(total_time / dt) + 1
        reflectivity = np.zeros(ns)
        for t, r in zip(self.times, self.coeffs):
            idx = int(t / dt)
            if idx < ns:
                reflectivity[idx] = r
        return reflectivity

class TraceGenerator:
    def __init__(self, wavelet_freq=30, dt=0.001, noise_std=0.0):
        self.dt = dt
        self.wavelet = RickerWavelet(wavelet_freq, dt).wavelet
        self.noise_std = noise_std
    
    def generate(self, reflectivity):
        trace = np.convolve(reflectivity, self.wavelet, mode='same')
        if self.noise_std > 0:
            trace += np.random.normal(0, self.noise_std, len(trace))
        return trace
```

```python
# Задаём модель: 3 слоя, 2 границы
depths = [500, 1200]           # м (глубины границ)
velocities = [2000, 2500, 3000]  # м/с (в слоях: выше 1й границы, между 1й и 2й, ниже 2й)
densities = [2100, 2200, 2400]   # кг/м³

# Параметры генерации
dt = 0.001          # 1 мс
total_time = 4.0    # 2 секунды
freq = 30           # Гц
noise = 0.01

# Создаём трассу
rc = ReflectionCoefficients(depths, velocities, densities)
reflectivity = rc.get_reflectivity_series(dt, total_time)
generator = TraceGenerator(wavelet_freq=freq, dt=dt, noise_std=noise)
trace = generator.generate(reflectivity)

# Временная ось
time_axis = np.arange(0, total_time + dt, dt)[:len(trace)]

# Визуализация
fig, axes = plt.subplots(3, 1, figsize=(12, 8))
axes[0].stem(rc.times, rc.coeffs, linefmt='r-', markerfmt='ro', basefmt='k-')
axes[0].set_title('Коэффициенты отражения')
axes[0].set_ylabel('R')

axes[1].plot(np.arange(-0.1, 0.1, dt), RickerWavelet(freq, dt).wavelet)
axes[1].set_title(f'Вейвлет Рикера, f={freq} Гц')
axes[1].set_ylabel('Амплитуда')

axes[2].plot(time_axis+1, trace)
for t in rc.times:
    axes[2].axvline(x=t, color='gray', linestyle='--', alpha=0.5)
axes[2].set_title('Синтетическая трасса (свёртка)')
axes[2].set_xlabel('Время (с)')
axes[2].set_ylabel('Амплитуда')
plt.tight_layout()
plt.show()
```

```python
def create_sgy_file(traces, dt_ms, filename="output.sgy"):
    """
    Создаёт SEG-Y файл из массива трасс.
    
    Параметры:
    traces    : 2D массив [num_traces x num_samples]
    dt_ms     : интервал дискретизации в миллисекундах
    filename  : имя выходного файла
    """
    num_traces, num_samples = traces.shape
    dt_us = int(dt_ms * 1000)  # переводим в микросекунды для SEG-Y
    
    # Создаём спецификацию
    spec = segyio.spec()
    spec.ilines = [1]  # один inline (прямая линия)
    spec.xlines = [num_traces]
    spec.samples = range(num_samples)
    spec.format = 5  # 4-байтовый IEEE float
    
    with segyio.create(filename, spec) as f:
        # Бинарный заголовок
        f.bin[segyio.BinField.Interval] = dt_us
        f.bin[segyio.BinField.Samples] = num_samples
        f.bin[segyio.BinField.Format] = 5
        
        # Запись трасс
        for i in range(num_traces):
            f.trace[i] = traces[i]
            
            # Базовые заголовки трасс
            f.header[i][segyio.TraceField.TraceNumber] = i + 1
            f.header[i][segyio.TraceField.TraceIdentifier] = 1  # живая трасса
    
    print(f"Файл {filename} создан: {num_traces} трасс, {num_samples} отсчётов, dt={dt_ms} мс")
    return filename

# Создаём несколько трасс с разной геологией (меняем глубину одного горизонта)
num_traces = 50
traces_array = []
depth_variation = np.linspace(-50, 50, num_traces)  # изменение глубины 2го горизонта от -50 до +50 м

for i in range(num_traces):
    # Меняем глубину второго горизонта
    depths_i = [500, 1200 + depth_variation[i]]
    velocities_i = [2000, 2500, 3000]
    densities_i = [2100, 2200, 2400]
    
    rc_i = ReflectionCoefficients(depths_i, velocities_i, densities_i)
    reflectivity_i = rc_i.get_reflectivity_series(dt, total_time)
    trace_i = generator.generate(reflectivity_i)
    traces_array.append(trace_i)

traces_array = np.array(traces_array)

# Сохраняем в SEG-Y
sgy_filename = "synth_section.sgy"
create_sgy_file(traces_array, dt_ms=1, filename=sgy_filename)

# Визуализация разреза (сейсмограмма)
plt.figure(figsize=(12, 6))
plt.imshow(traces_array.T, aspect='auto', cmap='seismic', 
           extent=[0, num_traces, total_time, 0])
plt.colorbar(label='Амплитуда')
plt.xlabel('Номер трассы')
plt.ylabel('Время (с)')
plt.title('Синтетический сейсмический разрез')
plt.show()
```

```python
def load_sgy_info(filename):
    """Загружает SEG-Y файл и возвращает трассы и параметры."""
    with segyio.open(filename, 'r', ignore_geometry=True) as f:
        # Читаем бинарный заголовок
        dt_us = f.bin[segyio.BinField.Interval]  # интервал в микросекундах
        dt_ms = dt_us / 1000
        num_samples = f.bin[segyio.BinField.Samples]
        num_traces = len(f.trace)
        
        # Читаем трассы
        traces = np.array([f.trace[i] for i in range(num_traces)])
        
        # Временная ось
        time_axis = np.arange(num_samples) * dt_ms / 1000  # в секундах
        
    print(f"Файл: {filename}")
    print(f"  Трасс: {num_traces}")
    print(f"  Отсчётов на трассу: {num_samples}")
    print(f"  Интервал: {dt_ms} мс ({dt_us} мкс)")
    print(f"  Формат: {f.bin[segyio.BinField.Format]}")
    
    return traces, time_axis, dt_ms

# Пример загрузки (замените на ваш файл)
# real_traces, real_time, real_dt = load_sgy_info("real_data.sgy")
```

```python
def compare_traces(synthetic, real, time_axis, title="Сравнение трасс"):
    """
    Сравнивает две трассы по нескольким параметрам.
    """
    # Выравниваем длины (если различаются)
    min_len = min(len(synthetic), len(real))
    synth = synthetic[:min_len]
    real_t = real[:min_len]
    time = time_axis[:min_len]
    
    # 1. Коэффициент корреляции Пирсона
    corr = np.corrcoef(synth, real_t)[0, 1]
    
    # 2. Коэффициент Спирмена (ранговая корреляция)
    spearman_corr, spearman_p = spearmanr(synth, real_t)
    
    # 3. MSE (средний квадрат ошибки)
    mse = np.mean((synth - real_t)**2)
    
    # 4. Нормированная MSE
    norm_mse = mse / np.var(real_t)
    
    print("=" * 50)
    print(f"РЕЗУЛЬТАТЫ СРАВНЕНИЯ")
    print("=" * 50)
    print(f"Коэффициент корреляции Пирсона: {corr:.4f}")
    print(f"Коэффициент корреляции Спирмена: {spearman_corr:.4f} (p={spearman_p:.4e})")
    print(f"MSE: {mse:.6f}")
    print(f"Нормированная MSE: {norm_mse:.4f}")
    
    # 5. Визуализация
    fig, axes = plt.subplots(2, 2, figsize=(12, 8))
    
    # Наложенные трассы
    axes[0, 0].plot(time, synth, 'b-', label='Синтетика', alpha=0.8)
    axes[0, 0].plot(time, real_t, 'r-', label='Реальная', alpha=0.8)
    axes[0, 0].set_title(f'Трассы (корр={corr:.3f})')
    axes[0, 0].set_xlabel('Время (с)')
    axes[0, 0].legend()
    axes[0, 0].grid(True)
    
    # Разность (ошибка)
    diff = synth - real_t
    axes[0, 1].plot(time, diff, 'g-')
    axes[0, 1].set_title(f'Разность (MSE={mse:.6f})')
    axes[0, 1].set_xlabel('Время (с)')
    axes[0, 1].grid(True)
    
    # Корреляционная функция (по времени)
    lags = np.arange(-min_len//4, min_len//4)
    corr_func = correlate(synth, real_t, mode='same') / (min_len * np.std(synth) * np.std(real_t))
    center = len(corr_func)//2
    axes[1, 0].plot(lags, corr_func[center - len(lags)//2 : center + len(lags)//2])
    axes[1, 0].axvline(x=0, color='k', linestyle='--')
    axes[1, 0].set_title('Взаимная корреляция')
    axes[1, 0].set_xlabel('Задержка (отсчёты)')
    axes[1, 0].grid(True)
    
    # Спектры
    freq = np.fft.rfftfreq(min_len, d=time[1]-time[0])
    spec_synth = np.abs(np.fft.rfft(synth))
    spec_real = np.abs(np.fft.rfft(real_t))
    axes[1, 1].plot(freq, spec_synth, 'b-', label='Синтетика', alpha=0.8)
    axes[1, 1].plot(freq, spec_real, 'r-', label='Реальная', alpha=0.8)
    axes[1, 1].set_title('Амплитудные спектры')
    axes[1, 1].set_xlabel('Частота (Гц)')
    axes[1, 1].set_ylabel('Амплитуда')
    axes[1, 1].legend()
    axes[1, 1].grid(True)
    axes[1, 1].set_xlim(0, 100)
    
    plt.tight_layout()
    plt.show()
    
    return {'pearson': corr, 'spearman': spearman_corr, 'mse': mse, 'norm_mse': norm_mse}

# Пример использования (для демонстрации сравниваем трассу с самой собой)
test_trace = traces_array[25]  # берём одну из сгенерированных
compare_traces(test_trace, test_trace, time_axis)
```

```python
def create_pseudo_well_logs():
    """Создаёт учебные каротажные кривые: глубины, скорости, плотности."""
    # Глубины (м)
    depths_log = np.linspace(0, 2000, 1000)
    
    # Искусственная скорость (м/с): тренд + колебания на границах слоёв
    velocity = 2000 + depths_log * 0.5  # линейный тренд
    # Добавляем слоистую структуру
    for depth, v_inc in [(500, +300), (800, -200), (1200, +400), (1600, -150)]:
        idx = np.argmin(np.abs(depths_log - depth))
        velocity[idx:idx+20] += v_inc
    
    # Плотность по формуле Гарднера (для песков и глин)
    density = 1.0 * velocity ** 0.25
    
    return depths_log, velocity, density

# Создаём псевдо-каротаж
depths_well, v_well, rho_well = create_pseudo_well_logs()

# Вычисляем импеданс
z_well = v_well * rho_well

# Визуализация
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
axes[0].plot(v_well, depths_well)
axes[0].set_title('Скорость V')
axes[0].set_xlabel('м/с')
axes[0].invert_yaxis()
axes[1].plot(rho_well, depths_well)
axes[1].set_title('Плотность ρ')
axes[1].set_xlabel('кг/м³')
axes[1].invert_yaxis()
axes[2].plot(z_well, depths_well)
axes[2].set_title('Акустический импеданс Z')
axes[2].set_xlabel('м/с * кг/м³')
axes[2].invert_yaxis()
plt.tight_layout()
plt.show()
```
